**Read the `sp500_companies.csv` dataset with spark and take the first column as header**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window

companies=spark.read.csv('sp500_companies.csv', header=True, sep=",")
companies.show(5)
```

**Define the schema for the company dataset `Schema_companies` and display it without inference**

```python
Schema_companies = StructType([
    StructField("Exchange",       StringType(),  True),
    StructField("Symbol",     StringType(),  True),
    StructField("Shortname",       StringType(),  True),
    StructField("Longname",       StringType(),  True),
    StructField("Sector",        StringType(),  True),
    StructField("Industry",      StringType(),  True),
    StructField("Currentprice",  FloatType(),  True),
    StructField("Marketcap",     LongType(),    True),
    StructField("Ebitda",     LongType(),    True),
    StructField("Revenuegrowth",     FloatType(),    True),
    StructField("City",     StringType(),    True),
    StructField("State",     StringType(),    True),
    StructField("Country",     StringType(),    True),
    StructField("Fulltimeemployees",     LongType(),    True),
    StructField("Longbusinesssummary",     StringType(),    True),
    StructField("Weight",     DoubleType(),    True)
])

# schema wi (without inference)
companies_wi= (spark.read
      .option("header", "true")
      .schema(Schema_companies)
      .csv("sp500_companies.csv"))
companies_wi.printSchema()
```

**Read the dataset `sp500_index.csv` in `index` using the first column as header**



```python
index=spark.read.csv('sp500_index.csv', header=True, sep=",")
index.show(5)
```

**Define the schema for the index dataset in `Schema_index` an display it with `index_wi` it without inference**



```python
Schema_index = StructType([
    StructField("Date",       DateType(),  True),
    StructField("S&P500",     FloatType(),  True)
])
# schema wi (without inference)
index_wi= (spark.read
      .option("header", "true")
      .schema(Schema_index)
      .csv("sp500_index.csv"))
index_wi.printSchema()
```

**Read the dataset `sp500_stocks.csv` in `index` using the first column as header**

```python
stocks=spark.read.csv('sp500_stocks.csv', header=True, sep=",")
stocks.show(5)
```

**Define the schema for the index dataset in `Schema_stocks` and display it with `stocks_wi` it without inference**



```python
Schema_stocks = StructType([
    StructField("Date",       DateType(),  True),
    StructField("Symbol",     StringType(),  True),
    StructField("Adj close",  DoubleType(),  True),
    StructField("Close",      DoubleType(),  True),
    StructField("High",       DoubleType(),  True),
    StructField("Low",        DoubleType(),  True),
    StructField("Open",       DoubleType(),  True),
    StructField("Volume",     LongType(),    True),
])
# schema wi (without inference)
stocks_wi= (spark.read
      .option("header", "true")
      .schema(Schema_stocks)
      .csv("sp500_stocks.csv"))

stocks_wi.printSchema()
```

**Erase in the dataset `companies.csv` the line with `NaN` or `None` as value and show it with `companies_clean`**

```python
companies_clean = companies_wi.dropna(subset=["Exchange","Symbol","Shortname","Longname","Sector","Industry",
"Currentprice","Marketcap","Ebitda","Revenuegrowth","City","State","Country","Fulltimeemployees","Longbusinesssummary","Weight"])
companies_clean.show(10)
```

**Erase in the dataset `index.csv` the line with `NaN` or `None` as value and show it with `index_clean`**

```python
index_clean=index_wi.dropna(subset=["Date","S&P500"])
index_clean.show(5)
```

**Erase in the dataset `sp500_stocks.csv` the line with `NaN` or `None` as value and show it with `index_stocks`**

```python
stocks_clean=stocks_wi.dropna(subset=["Date","Symbol","Adj close","Close","High","Low","Open","Volume"])
stocks_clean.show(5)
```

**This one create a column depending of the `Open` and `Close` value of the stocks in order to calculate the `Daily return`**

```python
daily_return = stocks_clean.withColumn("Daily_return", (col("Close") - col("Open")) / col("Open"))
daily_return.show(5)
```

**Just add a column `is_gain` from the table which show `1` when it's a gain and `0` when it's not.This is done by comparing if the value of the `Daily_return` is positive or not**

```python
is_gain = daily_return.withColumn("is_gain", (col("Daily_return")>0).cast("int"))
is_gain.show(10)
```

**Here I remove the "Adj Close" column from the the table beacause, it will not be useful for the following steps**

```python
remove_col=is_gain.drop(col("Adj Close"))
remove_col.show(10)
```

**Right here, I agregate data based their `symbol` and aggregate them according to the mean of the `daily_return` of those symbol as `symbole_return`**

```python
symbole_return =remove_col.groupBy("symbol").agg(avg("daily_return"))
symbole_return.show()
```

**Join the dataset `sp500_companies.csv` and `sp500_stocks.csv` according to the to match their value in each dataset.**

**And after group the companies by sector and aggregate them of the average of `Daily return` and named the column created under `average_daily_return`**

```python
# Effectuer la jointure
join_stocks = remove_col.join(companies_clean, on="Symbol", how="inner")

# Maintenant regrouper par secteur et calculer le rendement quotidien moyen
# Utiliser 'avg' directement tel qu'importé, et 'daily_return' est maintenant disponible dans joined_df
result = join_stocks.groupBy("Sector").agg(avg("Daily_return").alias("average_daily_return"))

result.show()
```

**Show the statistical in information of the column chosed such as `open`,`close`,`volume`,`daily_return`,`is_gain`**

```python
remove_col.describe(['open', 'close', 'volume', 'daily_return','is_gain']).show()
```

**Now, it show the filtering of the `volume` column for when volume greater than `10.000.000`**

```python
remove_col.filter(col("volume") > 10_000_000).show()
```

**Filter the dataset and displaying when `open` and `close` are greater than 0**

```python
aberant = remove_col.filter((col("open") > 0) & (col("close") > 0))
aberant.show()
```

**Show in the column `range`the range between the highest and the lowest value everyday**

```python
range= remove_col.withColumn("range", col("high") - col("low"))
range.show()
```

**Create a column `year` and `month` which take year, the month from the column `Date`,make an aggregation of several level with `rollup` and in the end aggregate the all according of the closing value of the stocks**



```python
year_close = range.withColumn("year", year(col("Date"))).withColumn("month", month(col("Date"))).rollup("year", "month", "symbol").agg(avg("close"))
year_close.show()
```

**Join `sector`,`Symbol` value with the funcion `cube` with all possibles combination (detail + sub‑totals sector + sub‑totals symbol + total), aggregate the value according to the `daily_return` and rename the column `average_daily_return`**

```python
joined_df_for_cube = range.join(companies_clean, on="Symbol", how="inner")
sector_symbol = joined_df_for_cube.cube("Sector", "Symbol").agg(avg("Daily_return").alias("average_daily_return"))
sector_symbol.show()
```

**Create a column `year` and `month` which take year, the month from the column `Date`,group them according to `year` and `month` and finally agregate column according of the sum of `volume`**



```python
volume_year = range.withColumn("year", year(col("Date"))).withColumn("month", month(col("Date"))).groupBy("year", "month").agg(sum("volume")).orderBy(col("year"), col("month"))
volume_year.show()
```

**Split the data by action (in other word treat each symbol independantly), sort according to the date by choosing only 20 lines starting from the current line**

```python
w = Window.partitionBy("symbol").orderBy("Date").rowsBetween(-19, 0)

partition = range.withColumn("sma20", avg("close").over(w))

partition.show()
```

**Use the `Pearson correlation` to see if the two column are correlated.**

**And as result mean that the relation between the both of them is not clear.**

```python
stats = partition.stat.corr("close", "volume")
print(stats)
```

**Calculate the quantile of the entry percentage `50%`,`90%`,`99%` and give as output an array with the same dimmension.**

```python
quantile = partition.approxQuantile("daily_return", [0.5, 0.9, 0.99], 0.01)
print(quantile)
```

**This is used to detect outlier in the dataset.**

**It create an `anomaly` column which display `1` when the `daily_return` is greater than `5%` return.**

```python
anomaly = partition.withColumn("anomaly", when(col("daily_return") > 0.05, 1).otherwise(0))
anomaly.show()
```

**Split the data into process it independantly for `january`,`february` and join all the result**

```python
partition_with_month = partition.withColumn("month", month(col("Date")))
jan = partition_with_month.filter(col("month") == 1)
feb = partition_with_month.filter(col("month") == 2)

union = jan.union(feb)
union.show()
```
